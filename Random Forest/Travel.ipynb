```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
df=pd.read_csv('Travel.csv')
```

```python
df.head()
```

```python
df['Gender']=df['Gender'].replace('Fe Male','Female')
```

```python
df['Gender'].value_counts()
```

```python
df['MaritalStatus']=df['MaritalStatus'].replace('Single','Unmarried')
```

```python
df['MaritalStatus'].value_counts()
```

```python
features_with_na=[features for features in df.columns if df[features].isnull().sum()>=1]
for features in features_with_na:
    print(features)
```

```python
df.Age.fillna(df.Age.median(),inplace=True)
df.TypeofContact.fillna(df.TypeofContact.mode()[0],inplace=True)
df.DurationOfPitch.fillna(df.DurationOfPitch.median(),inplace=True)
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode()[0],inplace=True)
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode()[0],inplace=True)
df.NumberOfTrips.fillna(0,inplace=True)
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode()[0],inplace=True)
df.MonthlyIncome.fillna(df.MonthlyIncome.median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop('CustomerID',inplace=True,axis=1)
```

```python
df['TotalVisiting']=df['NumberOfChildrenVisiting']+df['NumberOfPersonVisiting']
```

```python
df.drop(['NumberOfPersonVisiting','NumberOfChildrenVisiting'],inplace=True,axis=1)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(['ProdTaken'],axis=1)
y=df['ProdTaken']
```

```python
y.value_counts()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=10)
```

```python
cat_features=[feature for feature in X.columns if X[feature].dtype=='O']
```

```python
cat_features
```

```python
num_features=[feature for feature in X.columns if X[feature].dtype!='O']
```

```python
num_features
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer

num_transformation=StandardScaler()
oh_transformer=OneHotEncoder(drop='first')

preprocessor=ColumnTransformer(
    [
        ('OneHotEncoder',oh_transformer,cat_features),
        ('StandardScaler',num_transformation,num_features)
    ]
)
```

```python
X_train=preprocessor.fit_transform(X_train)
```

```python
X_test=preprocessor.transform(X_test)
```
