```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df=pd.read_csv('height_weight_dataset.csv')
```

```python
df.head()
```

```python
# find out the relation between the features
plt.scatter( df['Weight_kg'], df['Height_cm'] )
plt.xlabel('weight')
plt.ylabel('height')
plt.show()
```

```python
# finding correlation
df.corr()
```

```python
#seaborn
import seaborn as sns
sns.pairplot(df)
plt.show()
```

```python
# divide features in independent and dependent features


X=df[['Weight_kg']] #double bracket to get the dataframe ALWAYS remember the indenpendet feature should be a 2 dimension array
y=df['Height_cm'] # dependent features should be a series means in one dimension
```

```python
# train test split

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
# standardization

from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test=scaler.transform(X_test)
```

```python
# apply linear regression

from sklearn.linear_model import LinearRegression

regression=LinearRegression()
```

```python
regression.fit(X_train,y_train)
```

```python
print('cofficient or slope: ',regression.coef_)
print('intercept: ',regression.intercept_)
```

```python
# plot the training data best fit line

plt.scatter(X_train,y_train)
plt.plot(X_train,regression.predict(X_train))
plt.show()
```

```python
# prediction for test data

y_pred=regression.predict(X_test)
```

```python
# performance

from sklearn.metrics import mean_absolute_error,mean_squared_error

mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
```

```python
print(mse)
print(mae)
```

```python
# R square

from sklearn.metrics import r2_score

r_square=r2_score(y_test,y_pred)
```

```python
print(r_square)
```
